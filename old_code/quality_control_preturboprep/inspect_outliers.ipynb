```python
from pathlib import Path

import pandas as pd
```

```python
# dirpath = Path(__file__).parent.resolve()
dirpath = Path.cwd()
qc_csv_outlier_dir = dirpath / "qc_outliers"
qc_csv_files = list(qc_csv_outlier_dir.glob("*_outliers.csv"))
```

```python
# one example to read one csv file
qc_csv_file = qc_csv_files[0]
outliers = pd.read_csv(qc_csv_file)
# merge all outlier csv files
outliers = pd.concat(pd.read_csv(f) for f in qc_csv_files)
outliers.reset_index(drop=True, inplace=True)
```

```python
print(outliers.describe().T)
print(outliers.columns)
print(outliers["path"].nunique())
```

```python
print(outliers.groupby(["z_sigma"]).size().sort_values(ascending=False))
print(outliers.groupby(["iqr"]).size().sort_values(ascending=False))
print(outliers.groupby(["mad"]).size().sort_values(ascending=False))
print(outliers.groupby(["threshold_rule"]).size().sort_values(ascending=False))
print(outliers.groupby(["percentile"]).size().sort_values(ascending=False))
```

```python
# outliers[outliers["metric"] == "mad"].sort_values(by="value", ascending=False)
outliers[outliers["percentile"]].sort_values(by="value", ascending=False)
outliers[outliers["z_sigma"]].sort_values(by="value", ascending=False)
outliers[outliers["z_sigma"]]["path"].nunique()
outliers[outliers["z_sigma"]]["path"].nunique()
```

```python
# filter if threshold_rule or z_sigma is true
outliers_filtered = outliers[outliers["threshold_rule"] | outliers["z_sigma"]]
# in the path column, split with "/" and remove the 5 first parts
outliers_filtered["path"] = outliers_filtered["path"].apply(
    lambda path_str: "/".join(path_str.split("/")[6:])
)
# remove duplicates
outliers_filtered = outliers_filtered.drop_duplicates(subset=["path"])
outliers_filtered
# # save the filtered outliers to a csv file
# outliers_filtered.to_csv("filtered_outliers_before_preprocessing.csv", index=False)
# # save the paths to a text file
# outliers_filtered["path"].to_csv(
#     "filtered_outliers_paths_before_preprocessing.txt", index=False, header=False
# )
```
